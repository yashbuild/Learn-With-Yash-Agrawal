```python
def levenshtein_distance(word1, word2):
    len1, len2 = len(word1), len(word2)

    cache = {}

    def compute_distance(i, j):
        if (i, j) in cache:
            return cache[(i, j)]
        
        if i == 0:
            return j # insert j operation

        if j == 0:
            return i # delete i operation
            
        if word1[i-1] == word2[j-1]:
            result = compute_distance(i-1, j-1)
        else:
            insert_cost = compute_distance(i, j-1)
            delete_cost = compute_distance(i-1, j)
            replace_cost = compute_distance(i-1, j-1)
            result = 1 + min(insert_cost, delete_cost, replace_cost)

        cache[(i, j)] = result
        return result
            
    return compute_distance(len1, len2)
```

```python
levenshtein_distance("cat", "bats")
```

```python
def get_best_spelling(word, eng_dict, max_suggestions=3):
    suggestions = {}

    for dict_word in eng_dict:
        score = levenshtein_distance(word, dict_word)
        suggestions[dict_word] = score

    # Sort by score (edit distance), then alphabetically
    sorted_suggestions = sorted(suggestions.items(), key=lambda x: (x[1], x[0]))

    # Return top N suggestions
    return [w for w in sorted_suggestions[:max_suggestions]]
```

```python
eng_dict = set([
    "natural", "neutral", "nation", 
    "processing", "progressing", "professing", 
    "machine", "mashing", "marine", 
    "love", "i", "language", "and", "learning"
])

input_sentence = "i love natral language progresing and mashine learning"

miss_spellings = {}
for word in input_sentence.split():
    if word not in eng_dict:
        miss_spellings[word] = get_best_spelling(word, eng_dict)

print(miss_spellings)
```
